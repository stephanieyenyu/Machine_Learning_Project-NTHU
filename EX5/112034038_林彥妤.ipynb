```python
import os
import pandas as pd
import tensorflow as tf
from ckiptagger import WS, POS, NER
from google.colab import drive
drive.mount('/content/drive')
```

```python
path = ('/content/drive/MyDrive/機器學習概論115/data/data')
```

```python
ws = WS(path)
pos = POS(path)
ner = NER(path)
```

```python
sentence_list = [
    "中央氣象署表示，今天受到東北季風影響，北部及東北部天氣較涼。",
    "中南部地區維持多雲到晴的天氣，午後山區有局部短暫陣雨。",
    "提醒民眾早晚溫差大，外出請適時添加衣物以免受涼。"
]

# 將句子中可能出現的空白字元清除，以免影響斷詞結果
sentence_list = [sentence.replace(" ", "") for sentence in sentence_list]
```

```python
# 斷詞
word_sentence_list = ws(
    sentence_list,
    # sentence_segmentation = True, # To consider delimiters 分隔符號
    # segment_delimiter_set = {",", "。", ":", "?", "!", ";"}), # This is the defualt set of delimit$
    # recommend_dictionary = dictionary1, # words in this dictionary are encouraged
    # coerce_dictionary = dictionary2, # words in this dictionary are forced
)

# 詞性標註
pos_sentence_list = pos(word_sentence_list)

# 實體
entity_sentence_list = ner(word_sentence_list, pos_sentence_list)
```

```python
print('WS: ', word_sentence_list)
print('POS: ', pos_sentence_list)
print('NER: ', entity_sentence_list)

for name in entity_sentence_list[2]:
    print(name)
```
