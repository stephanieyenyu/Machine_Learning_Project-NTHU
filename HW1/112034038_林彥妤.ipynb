# Setup & Import File

```python
#直接執行以下程式
import pandas as pd

# 1. 定義檔案 ID (助教提供的 garments_worker_productivity.csv)
file_id = '16lYjDveb2tkQ8X2Idf3s-ytjXhGbeAGp'
url = f'https://drive.google.com/uc?id={file_id}'

# 2. 讀取資料 (確保這一行左邊完全沒有空格)
data = pd.read_csv(url)

# 3. 顯示前五筆資料
display(data.head())
```

# Question 1

```python
# Show the datatype of every columns.
print("--- Q1: 資料類型 ---")
data.info()
```

# Question 2

```python
# Check each row for the number of NaN cells and show the column name and corresponding number. After that fill in the missing values in the 'wip' column with 0.
print("--- Q2: 填補缺失值 ---")
print("填補前缺失值數量：\n", data.isnull().sum())

data['wip'] = data['wip'].fillna(0)

print("\n填補後缺失值數量：\n", data.isnull().sum())
```

# Question 3

```python
# Mean and median of targeted productivity.
mean_val = data['targeted_productivity'].mean()
median_val = data['targeted_productivity'].median()

print("--- Q3: targeted_productivity 統計 ---")
print(f"Mean = {mean_val}")
print(f"Median = {median_val}")
```

# Question 4

```python
# Find the correlation of team, idle time, no of workers, and targeted productivity.
cols = ['team', 'idle_time', 'no_of_workers', 'targeted_productivity']
corr_matrix = data[cols].corr()

display(corr_matrix)
```

# Question 5

```python
# Save the standardized 'actual_productivity' values to a new column, 'Z_actual', and then display the first and last 5 rows after removing 'targeted_productivity'.
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
data['Z_actual'] = scaler.fit_transform(data[['actual_productivity']])
data = data.drop(columns=['targeted_productivity'], errors='ignore')
display(data)
```

# Question 6

```python
# 將smv分成不同的區間，分別是'0-10', '11-20', '21-30', '31-40', '41-50', '50+'，然後將每一列對應的區間列出
bins = [0, 10, 20, 30, 40, 50, float('inf')]
labels = ['0-10', '11-20', '21-30', '31-40', '41-50', '50+']
data['smv_group'] = pd.cut(data['smv'], bins=bins, labels=labels, include_lowest=True)
display(data[['smv', 'smv_group']])
```

# Question 7

```python
# 將'quarter'column 進行季度映射，使用{'Quarter1': 1, 'Quarter2': 2, 'Quarter3': 3, 'Quarter4': 4, 'Quarter5': 5}進行標籤編碼，並将結果儲存到新column'quarter_encoded'中。
quarter_map = {
    'Quarter1': 1,
    'Quarter2': 2,
    'Quarter3': 3,
    'Quarter4': 4,
    'Quarter5': 5
}
data['quarter_encoded'] = data['quarter'].map(quarter_map).astype(float)
display(data)
```

# Question 8

```python
# Export csv
filename = '112034038_林彥妤.csv'
data.to_csv(filename, index=False)

print(f"--- Q8: {filename} ---")
```
