# Import Packages

```python
import sklearn
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from google.colab import drive
from sklearn.model_selection import train_test_split
from sklearn import metrics
```

# Data Preprocessing

```python
##讀檔:可用自己習慣的方式讀檔
from google.colab import files
data = files.upload()
```

```python
# Read file
path = '/content/Heart Failure Clinical Records.csv'
data = pd.read_csv(path)
```

```python
# Drop Label on X
X = data.drop(['time', 'DEATH_EVENT'], axis=1).copy()
X
```

```python
# Assign target
y = data['DEATH_EVENT'].copy()
y
```

```python
# Splitting into train and test (test_size=0.25, random_state=2)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=2)
```

# Train

```python
from sklearn.linear_model import LogisticRegression
from xgboost import XGBClassifier
from sklearn import neighbors
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

lr_model = LogisticRegression(max_iter=1000)
lr_model.fit(X_train, y_train)

xgb_model = XGBClassifier(use_label_encoder=False, eval_metric='logloss')
xgb_model.fit(X_train, y_train)

dt_model = DecisionTreeClassifier(random_state=2)
dt_model.fit(X_train, y_train)

rf_model = RandomForestClassifier(random_state=2)
rf_model.fit(X_train, y_train)
```

# Prediction

```python
lr_pred = lr_model.predict(X_test)
xgb_pred = xgb_model.predict(X_test)
dt_pred = dt_model.predict(X_test)
rf_pred = rf_model.predict(X_test)

print(lr_pred)
```

# F1_Score

```python
from sklearn.metrics import f1_score
print("LogisticRegression F1-Score:", f1_score(y_test, lr_pred))
print("XGBoost F1-Score:", f1_score(y_test, xgb_pred))
print("Decision Tree F1-Score:", f1_score(y_test, dt_pred))
print("RandomForest F1-Score:", f1_score(y_test, rf_pred))
```
