Import packages

```python
import numpy as np
from sklearn.manifold import TSNE
from sklearn import datasets
import matplotlib.pyplot as plt
```

Load data from sklearn

X : feature of the digit

y : label of the digit

```python
digits = datasets.load_digits(n_class=6)
X, y = digits.data, digits.target
```

Check the shape of the data

(number of data, number of feateures)

```python
X.shape
```

Reshape the data to see the origin digit

```python
i = np.random.randint(0,1083)
img = X[i,:].reshape((8,8))
plt.title(y[i])
plt.imshow(img)
```

Use t-SNE to reduce the dimension to 2 and plot the data

```python
tsne = TSNE(n_components=2, random_state=42)
reduced_X = tsne.fit_transform(X)
plt.scatter(reduced_X[:, 0], reduced_X[:, 1], c=y)
plt.show()
```
